### **1. Sales Revenue & Order Value Distribution (Central Tendency & Spread)**
- TABLE 1: Order-level financial metrics
- TABLE 2: Descriptive stats for summarized financial metrics
- GRAPH 1: Distribution of Order Revenue
- GRAPH 2: Distribution of Freight Costs

```python
# Query to pull order-level financial metrics
query = """
    SELECT 
        f.order_id,
        SUM(f.unit_price * f.quantity) AS order_revenue,
        MAX(f.freight) AS freight_cost,
        SUM(f.unit_price * f.quantity) + MAX(f.freight) AS total_order_amount
    FROM dw.fact_sales f
    GROUP BY f.order_id;
"""

# Loading into a pandas DataFrame
df_orders = pd.read_sql(query, con=dw_engine)

print("--- TABLE 1: Order-level financial metrics ---")
display(df_orders.head())
```

```python
# Query to get descriptive statistics
def calculate_advanced_stats(series, name="Metric"):
    """Helper function to compute detailed descriptive stats including IQR and Skew."""
    stats_dict = {
        "Metric": name,
        "Mean": round(series.mean(), 2),
        "Median": round(series.median(), 2),
        "Std Dev": round(series.std(), 2),
        "Min": round(series.min(), 2),
        "Q1 (25%)": round(series.quantile(0.25), 2),
        "Q3 (75%)": round(series.quantile(0.75), 2),
        "IQR": round(series.quantile(0.75) - series.quantile(0.25), 2),
        "Max": round(series.max(), 2),
        "Skewness": round(series.skew(), 2)
    }
    return stats_dict

# Compute and display summary table
stats_summary = pd.DataFrame([
    calculate_advanced_stats(df_orders["order_revenue"], "Order Revenue ($)"),
    calculate_advanced_stats(df_orders["freight_cost"], "Freight Cost ($)"),
    calculate_advanced_stats(df_orders["total_order_amount"], "Total Order Amount ($)")
])

print("--- TABLE 2: Descriptive stats for summarized financial metrics ---")
display(stats_summary.set_index("Metric"))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Order Revenue Distribution
sns.histplot(df_orders["order_revenue"], kde=True, ax=axes[0], color="royalblue", bins=30)
axes[0].set_title("Distribution of Order Revenue (Right-Skewed)")
axes[0].axvline(df_orders["order_revenue"].median(), color='red', linestyle='--', label=f'Median: ${df_orders["order_revenue"].median():,.2f}')
axes[0].axvline(df_orders["order_revenue"].mean(), color='green', linestyle='-', label=f'Mean: ${df_orders["order_revenue"].mean():,.2f}')
axes[0].legend()

# Plot 2: Freight Cost Distribution
sns.histplot(df_orders["freight_cost"], kde=True, ax=axes[1], color="darkorange", bins=30)
axes[1].set_title("Distribution of Freight Costs")
axes[1].axvline(df_orders["freight_cost"].median(), color='red', linestyle='--', label=f'Median: ${df_orders["freight_cost"].median():,.2f}')
axes[1].axvline(df_orders["freight_cost"].mean(), color='green', linestyle='-', label=f'Mean: ${df_orders["freight_cost"].mean():,.2f}')
axes[1].legend()

plt.tight_layout()
print("--- GRAPH 1 and 2: Distribution of Order Revenue and Freight Costs ---")
plt.show()
```

   - **Order Revenue:**

   - **Freight Costs:**

### **2. Product Performance Breakdown & Pareto Analysis**
- TABLE 3: Product Performance and Stock Metrics
- GRAPH 3: Top 10 Revenue-Generating Products
- TABLE 4: Dead Stock Items (High Stock, Low Sales)

```python
# Query for product performance and stock metrics
product_query = """
    SELECT 
        p.product_name,
        c.category_name,
        p.units_in_stock,
        SUM(f.quantity) AS total_units_sold,
        SUM(f.unit_price * f.quantity) AS total_revenue
    FROM dw.fact_sales f
    JOIN dw.dim_products p ON f.product_id_sk_fk = p.product_id_sk_pk
    JOIN dw.dim_categories c ON p.category_id_sk_fk = c.category_id_sk_pk
    GROUP BY p.product_name, c.category_name, p.units_in_stock
    ORDER BY total_revenue DESC;
"""

df_products = pd.read_sql(product_query, con=dw_engine)

# Calculate Stock Turnover Ratio (handle division by zero if stock is 0)
df_products["stock_turnover"] = np.where(
    df_products["units_in_stock"] > 0, 
    df_products["total_units_sold"] / df_products["units_in_stock"], 
    np.nan
)

print("--- TABLE 3: Product Performance and Stock Metrics  ---")
display(df_products.head())
```

```python
# Creating bar chart
plt.figure(figsize=(10, 6))
top_products = df_products.head(10).sort_values(by="total_revenue", ascending=True)

sns.barplot(
    data=top_products, 
    x="total_revenue", 
    y="product_name", 
    hue="category_name", 
    palette="Blues_r", 
    dodge=False
)

plt.title("Top 10 Revenue-Generating Products", fontsize=14, fontweight="bold")
plt.xlabel("Total Revenue ($)", fontsize=12)
plt.ylabel("Product Name", fontsize=12)
plt.legend(title="Category", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
print("--- GRAPH 3: Top 10 Revenue-Generating Products ---")
plt.show()
```

```python
# Sort products by revenue descending
df_pareto = df_products.sort_values(by="total_revenue", ascending=False).reset_index(drop=True)

# Calculate cumulative revenue percentage
df_pareto["cumulative_revenue"] = df_pareto["total_revenue"].cumsum()
total_store_revenue = df_pareto["total_revenue"].sum()
df_pareto["cumulative_percentage"] = (df_pareto["cumulative_revenue"] / total_store_revenue) * 100

# Find how many products make up 80% of the revenue
products_driving_80 = df_pareto[df_pareto["cumulative_percentage"] <= 80]
percentage_of_catalog = (len(products_driving_80) / len(df_pareto)) * 100

print(f"Total Catalog Size: {len(df_pareto)} products")
print(f"Products driving 80% of revenue: {len(products_driving_80)} ({percentage_of_catalog:.1f}% of total catalog)")
```

```python
# Filter for dead stock: High units in stock, but low or zero units sold
dead_stock = df_products.sort_values(by=["units_in_stock", "total_units_sold"], ascending=[False, True])

print("--- TABLE 4: Dead Stock Items (High Stock, Low Sales) ---")
display(dead_stock.head(10))
```

### **3. Customer Concentration & Purchasing Frequency**
- TABLE 5: Customer Purchasing Frequency & Lifetime Value
- TABLE 6: Customers with Highest Inactivity (Potential Churn Risk)
- GRAPH 4: Top 10 Customers by Total Spend (Bar chart)

```python
# 1. Query customer-level transaction details
customer_query = """
    SELECT 
        c.company_name,
        f.order_id,
        d.full_date AS order_date,
        SUM(f.unit_price * f.quantity) AS order_revenue
    FROM dw.fact_sales f
    JOIN dw.dim_customers c ON f.customer_id_sk_fk = c.customer_id_sk_pk
    JOIN dw.dim_date d ON f.order_date_id_sk_fk = d.date_id_sk_pk
    GROUP BY c.company_name, f.order_id, d.full_date;
"""

df_customer_orders = pd.read_sql(customer_query, con=dw_engine)

# Convert order_date to datetime format
df_customer_orders["order_date"] = pd.to_datetime(df_customer_orders["order_date"])

# 2. Aggregate total spend and order frequency per customer
df_customer_summary = df_customer_orders.groupby("company_name").agg(
    total_spend=("order_revenue", "sum"),
    order_frequency=("order_id", "nunique"),
    first_order=("order_date", "min"),
    last_order=("order_date", "max")
).reset_index()

# 3. Calculate average time gap between repeat orders per customer
df_sorted = df_customer_orders.sort_values(by=["company_name", "order_date"])
df_sorted["days_between_orders"] = df_sorted.groupby("company_name")["order_date"].diff().dt.days

df_gaps = df_sorted.groupby("company_name")["days_between_orders"].mean().reset_index(name="avg_days_between_orders")

# Merge the average time gap back into our customer summary
df_customer_summary = df_customer_summary.merge(df_gaps, on="company_name", how="left")

# Sort by top spending VIP customers
df_customer_summary = df_customer_summary.sort_values(by="total_spend", ascending=False).reset_index(drop=True)

print("--- TABLE 5:  Customer Purchasing Frequency & Lifetime Value---")
display(df_customer_summary.head(10))
```

```python
# Bar Chart: Top 10 VIP Customers
plt.figure(figsize=(10, 6))
top_customers = df_customer_summary.head(10).sort_values(by="total_spend", ascending=True)

sns.barplot(
    data=top_customers,
    x="total_spend",
    y="company_name",
    hue="company_name",  # Added to prevent modern Seaborn warnings
    palette="viridis",
    legend=False
)

plt.title("Top 10 Customers by Total Spend (VIP Accounts)", fontsize=14, fontweight="bold")
plt.xlabel("Total Spend ($)", fontsize=12)
plt.ylabel("Customer Company Name", fontsize=12)
plt.tight_layout()
print("--- GRAPH 4: Top 10 Customers by Total Spend---")
plt.show()
```

```python
# Churn Risk Analysis
# Find the dataset's latest order date as a baseline for "today"
latest_dataset_date = df_customer_orders["order_date"].max()

# Calculate days since their last order
df_customer_summary["days_since_last_order"] = (latest_dataset_date - df_customer_summary["last_order"]).dt.days

# Filter for customers who haven't ordered in a while (longest inactive first)
churn_risk = df_customer_summary.sort_values(by="days_since_last_order", ascending=False)

print("--- TABLE 6: Customers with Highest Inactivity (Potential Churn Risk) ---")
display(churn_risk.head(10))
```

### **3. CONCLUSIONS for Customer Concentration & Purchasing Frequency**

   **3.1. Explanations**

     * Our smallest customer generated just **$140.00**, while our top VIP customer spent over **$115,000.00**!
     * The middle 50% of our customers (the IQR) range between **$2,100.00** and **$10,850.50** in total spend.

### **4. Employee Sales Productivity Ranking**
- TABLE 7: Employee Sales Productivity Summary
- GRAPH 5: Sales Representative Productivity: Total Revenue by Employee

```python
# 4. Employee Sales Productivity Ranking
# 1. Query employee-level sales details from the data warehouse
employee_query = """
    SELECT 
        CONCAT(e.first_name, ' ', e.last_name) AS employee_name,
        f.order_id,
        (f.unit_price * f.quantity) AS line_revenue,
        f.discount
    FROM dw.fact_sales f
    JOIN dw.dim_employees e ON f.employee_id_sk_fk = e.employee_id_sk_pk;
"""

df_employee_sales = pd.read_sql(employee_query, con=dw_engine)

# 2. Aggregate total revenue, order volume, and average discount per employee
df_employee_summary = df_employee_sales.groupby("employee_name").agg(
    total_revenue=("line_revenue", "sum"),
    total_orders_handled=("order_id", "nunique"),
    avg_discount_given=("discount", "mean")
).reset_index()

# Sort by top-performing revenue drivers
df_employee_summary = df_employee_summary.sort_values(by="total_revenue", ascending=False).reset_index(drop=True)

print("--- TABLE 7: Employee Sales Productivity Summary ---")
display(df_employee_summary.head(10))
```

```python
plt.figure(figsize=(10, 6))
top_employees = df_employee_summary.sort_values(by="total_revenue", ascending=True)

sns.barplot(
    data=top_employees,
    x="total_revenue",
    y="employee_name",
    hue="employee_name",
    palette="magma",
    legend=False
)

plt.title("Sales Representative Productivity: Total Revenue by Employee", fontsize=14, fontweight="bold")
plt.xlabel("Total Revenue Generated ($)", fontsize=12)
plt.ylabel("Employee Name", fontsize=12)
plt.tight_layout()
print("--- GRAPH 5: Sales Representative Productivity---")
plt.show()
```

### **Hypothesis Test:** 
**1st Hypothesis: Impact of Discounts on Order Volume**

* **Business Question:** Does granting a discount lead to significantly larger order quantities, or are sales reps giving away margin without driving higher volume?   
* **Data Context:** Order line items containing quantities and discount metrics linked across the warehouse tables (`fact_sales` and related dimension tables).

* **Hypotheses:**
  * **$H_0$ (Null Hypothesis):** There is no difference in the mean order quantity between orders with a discount (`discount > 0`) and orders with no discount (`discount = 0`).
  * **$H_1$ (Alternative Hypothesis):** Orders with a discount have a significantly higher mean order quantity than non-discounted orders.

```python
from scipy import stats

# Split order line items based on whether a discount was applied
discounted_orders = df_employee_sales[df_employee_sales["discount"] > 0]["line_revenue"]
no_discount_orders = df_employee_sales[df_employee_sales["discount"] == 0]["line_revenue"]

# Run the independent two-sample t-test
t_stat, p_val = stats.ttest_ind(discounted_orders, no_discount_orders, equal_var=False)
print(f"T-statistic: {t_stat:.4f}, P-value: {p_val:.4f}")
```

### **1st Hypothesis Test Result: Impact of Discounts on Order Revenue**

* **Statistical Findings:** 
  * **T-Statistic:** $3.1721$
  * **P-Value:** $0.0015$ ($p < 0.05$)
* **Conclusion:** We reject the null hypothesis. There is a statistically significant difference in revenue between discounted and non-discounted order lines, with discounted items showing a higher average revenue.
* **Business Insight:** Discounts are not merely eroding profit margins; they are statistically correlated with larger transaction sizes. This indicates that sales representatives are effectively leveraging discounts as incentives to drive higher-volume orders.

### **2nd Hypothesis: Top Tier vs. Regular Customer Spending**

* **Business Question:** Top VIP accounts (like *QUICK-Stop* and *Save-a-lot Markets*) generating a mean order value that is statistically higher than the rest of the customer base, or is their high total spend solely a result of placing more frequent orders?   

* **Hypotheses:**
  * **$H_0$ (Null Hypothesis):** The mean order revenue for top-tier customers (Top 10 by spend) is equal to the mean order revenue of regular customers.
  * **$H_1$ (Alternative Hypothesis):** Top-tier customers have a significantly higher mean order revenue per transaction.

```python
# Identify the Top 10 customers based on total spend
top_customers_list = df_customer_summary.head(10)["company_name"].tolist()

# Split order revenues into top-tier vs regular customer groups
top_tier_orders = df_customer_orders[df_customer_orders["company_name"].isin(top_customers_list)]["order_revenue"]
regular_orders = df_customer_orders[~df_customer_orders["company_name"].isin(top_customers_list)]["order_revenue"]

# Perform the Independent Two-Sample T-Test (Welch's T-Test with equal_var=False)
t_stat, p_val = stats.ttest_ind(top_tier_orders, regular_orders, equal_var=False)

# Print the results
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_val:.4f}")
```

### **2nd Hypothesis Test Result: Top-Tier vs. Regular Customer Revenue**

- **Statistical Findings:** 
  * **T-Statistic:** $8.8446$
  * **P-Value:** $0.0000$ ($p < 0.05$)
- **Conclusion:** We reject the null hypothesis. There is a statistically significant difference in order revenue between top-tier VIP customers and regular customers, with top accounts generating higher per-transaction revenue.
- **Business Insight:** Top-tier accounts do not merely accumulate high total spend through frequent ordering; they consistently drive higher revenue per individual transaction. This validates strategic investments in VIP account management and tailored corporate perks.
